**2.1** A researcher measures the weight of a standard object multiple times using the same scale under controlled conditions. However, each measurement varies slightly.

A: Random measurement error, a random error, the variation is small and inconsistent between measurements. There is no consistent shift in one direction, so this is random fluctuation due to instrument precision limits.

**2.2** A political pollster conducts a survey to gauge public opinion on a particular candidate's popularity. The pollster only calls landline telephone numbers during daytime hours to conduct the survey.

A: Selection bias, a systematic error, the sample systematically excludes people without landlines and people not home during the day (e.g., younger or working individuals).

**2.3** A university sends out a survey to all graduating students to gather feedback on their educational experience. Only 30% of the students respond to the survey.

A: Non-response bias, a systematic error, those who respond may differ systematically from those who do not (e.g., very satisfied or very dissatisfied students), skewing results.

2.4 A land survey is conducted with a laser rangefinder to compute distances.  It is later found that the laser rangefinder was miscalibrated resulting in all measurements being off by a factor 1.2.

A: Measurement bias, a systematic error, every measurement is consistently off by the same factor. This creates a consistent directional error.

2.5 A pharmaceutical company conducts a clinical trial to test the efficacy of a new drug. The company recruits participants from its own employees to participate in the trial.

A: Selection bias, systematic error, employees may not represent the broader population (health status, motivation, incentives), so results may not generalize.

2.6 A nurse measures patients' blood pressure using an automated blood pressure monitor. However, the blood pressure readings for each patient vary slightly with repeated measurements.

A: Random measurement error, random error, small natural fluctuations and device variability create inconsistent variation, not a consistent directional shift.

2.7 A police officer patrols a neighborhood with a reputation for criminal activity. The officer is aware of the neighborhood's reputation and characterizes neighborhood residents accordingly.

A: Observer bias, systematic error, the officer’s expectations influence interpretation and judgment, creating a consistent skew in perception.

2.8 A researcher studies the effectiveness of a new weight-loss program by asking participants to self-report their weight loss over the past month.

A: Self-reporting bias and social desirability bias, systematic eror, participants may overstate weight loss to appear successful or socially acceptable. This tends to systematically overestimate results.

2.9 A nutritional study assesses participants' weight loss over a 12-week period among participants at a weight loss camp.  There are two scales.  One is a newer, high precision digital scale and some use an older mechanical scale.  During the trial, the investigator running the trial fails to record which participants use which scale.

A: Measurement bias and random measurement error, random and systematic error possible, the newer scale may systematically differ from the older one (systematic bias). Since scale use isn’t tracked, variation between instruments may also introduce unexplained variability (appearing random in analysis)

2.10 NewShy is a potential treatment to reduce the occurrence of chronic headaches. The

## Part 5: Z-scores, Phi, Erf, and Gaussians

(10 points for this part)

5.1 You are given the scores of students in a national standardized math test. The mean score is 500 with a standard deviation of 100. John scored 620 on this test. Calculate John's Z score, and then determine the probability of a student scoring less than John. Show how you came to your result.  Practice using LaTeX.  (5 points)

A: $$
\mu = 500,\quad \sigma = 100,\quad x = 620
$$

$$
z = \frac{x-\mu}{\sigma}
     = \frac{620-500}{100}
     = \frac{120}{100}
     = 1.2
$$

$$
P(X < 620) = P(Z < 1.2) = \Phi(1.2)
$$

$$
\Phi(1.2) \approx 0.884
$$

5.2 A manufacturing process produces steel rods that are supposed to be 100 cm in length. However, due to variability in the process, the lengths of the rods follow a Gaussian distribution with a mean of 100 cm and a standard deviation of 2 cm. What is the probability that a randomly selected steel rod from this process will be either shorter than 96 cm or longer than 104 cm? Use the error function (erf) to compute this probability. Show in 
steps how you came to this answer.  Practice using LaTeX.  (5 points)

A: $$
X \sim \mathcal{N}(100,2^2)
$$

$$
\mu = 100,\quad \sigma = 2
$$

$$
z_{96} = \frac{96-100}{2} = \frac{-4}{2} = -2
$$

$$
z_{104} = \frac{104-100}{2} = \frac{4}{2} = 2
$$

$$
P(X<96 \ \text{or}\ X>104) = P(Z<-2) + P(Z>2)
$$

$$
P(Z<-2) + P(Z>2) = 2P(Z>2) = 2\left(1-\Phi(2)\right)
$$

$$
\Phi(z) = \frac{1}{2}\left(1+\mathrm{erf}\left(\frac{z}{\sqrt{2}}\right)\right)
$$

$$
\Phi(2) = \frac{1}{2}\left(1+\mathrm{erf}\left(\frac{2}{\sqrt{2}}\right)\right)
       = \frac{1}{2}\left(1+\mathrm{erf}\left(\sqrt{2}\right)\right)
$$

$$
P(X<96 \ \text{or}\ X>104)
= 2\left(1-\Phi(2)\right)
= 2\left(1-\frac{1}{2}\left(1+\mathrm{erf}(\sqrt{2})\right)\right)
$$

$$
= 2\left(\frac{1}{2}\left(1-\mathrm{erf}(\sqrt{2})\right)\right)
= 1-\mathrm{erf}(\sqrt{2})
$$

$$
P(X<96 \ \text{or}\ X>104)
= 2(1-0.97725)
= 2(0.02275)
= 0.0455
$$



```python
# PART 6 ANSWERSSS

import numpy as np
import matplotlib.pyplot as plt

# Reproducibility
rng = np.random.default_rng(42)

# Log-normal parameters (log-space)
mu = 0.0
sigma = 0.5

def lognormal_pdf(x, mu=0.0, sigma=0.5):
    x = np.asarray(x)
    pdf = np.zeros_like(x, dtype=float)
    mask = x > 0
    xm = x[mask]
    pdf[mask] = (1.0 / (xm * sigma * np.sqrt(2*np.pi))) * np.exp(-((np.log(xm) - mu)**2) / (2*sigma**2))
    return pdf

# X-range for PDF
x = np.linspace(1e-4, 5.0, 2000)
pdf = lognormal_pdf(x, mu, sigma)

# ---------- 6.1 ----------
m = 100   # number of sample means
n = 5     # samples per mean

samples_6_1 = rng.lognormal(mean=mu, sigma=sigma, size=m*n)         # all raw samples (500)
means_6_1 = samples_6_1.reshape(m, n).mean(axis=1)                  # 100 sample means

# Plot PDF + histogram of raw samples (relative frequency via density=True)
plt.figure()
plt.plot(x, pdf, linewidth=2)
plt.hist(samples_6_1, bins=25, density=True, alpha=0.5)
plt.title("6.1 Log-normal PDF with relative frequency histogram of raw samples (m=100, n=5)")
plt.xlabel("x")
plt.ylabel("density")
plt.xlim(0, 5.0)
plt.tight_layout()
plt.show()

# Plot sample means on the real line (rug/strip plot)
plt.figure()
plt.scatter(means_6_1, np.zeros_like(means_6_1), alpha=0.8)
plt.yticks([])
plt.title("6.1 Sample means on the real line (m=100, n=5)")
plt.xlabel("sample mean")
plt.tight_layout()
plt.show()

# ---------- 6.2 ----------
m = 10000
n = 5

samples_6_2 = rng.lognormal(mean=mu, sigma=sigma, size=m*n)
means_6_2 = samples_6_2.reshape(m, n).mean(axis=1)

plt.figure()
plt.plot(x, pdf, linewidth=2)
plt.hist(means_6_2, bins=60, density=True, alpha=0.5)
plt.title("6.2 Log-normal PDF with relative frequency histogram of sample means (m=10000, n=5)")
plt.xlabel("x")
plt.ylabel("density")
plt.xlim(0, 5.0)
plt.tight_layout()
plt.show()

# ---------- 6.3 ----------
m = 1000
n = 10

samples_6_3 = rng.lognormal(mean=mu, sigma=sigma, size=m*n)
means_6_3 = samples_6_3.reshape(m, n).mean(axis=1)

plt.figure()
plt.plot(x, pdf, linewidth=2)
plt.hist(means_6_3, bins=50, density=True, alpha=0.5)
plt.title("6.3 Log-normal PDF with relative frequency histogram of sample means (m=1000, n=10)")
plt.xlabel("x")
plt.ylabel("density")
plt.xlim(0, 5.0)
plt.tight_layout()
plt.show()

# ---------- 6.4 ----------
m = 1000
n = 20

samples_6_4 = rng.lognormal(mean=mu, sigma=sigma, size=m*n)
means_6_4 = samples_6_4.reshape(m, n).mean(axis=1)

plt.figure()
plt.plot(x, pdf, linewidth=2)
plt.hist(means_6_4, bins=50, density=True, alpha=0.5)
plt.title("6.4 Log-normal PDF with relative frequency histogram of sample means (m=1000, n=20)")
plt.xlabel("x")
plt.ylabel("density")
plt.xlim(0, 5.0)
plt.tight_layout()
plt.show()
```
